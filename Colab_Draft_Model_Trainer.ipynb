```python
# 3. Run 70,000 Sample High-Speed Training Pipeline
import math, time, os, torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer
from datasets import load_dataset

# Clear GPU cache
torch.cuda.empty_cache()

# Draft Transformer Definition
class RMSNorm(nn.Module):
    def __init__(self, dim, eps=1e-5):
        super().__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(dim))
    def forward(self, x):
        return x * torch.rsqrt(x.pow(2).mean(-1, keepdim=True) + self.eps) * self.weight

class RotaryEmbedding(nn.Module):
    def __init__(self, dim, max_seq_len=2048, base=500000.0):
        super().__init__()
        self.dim = dim
        inv_freq = 1.0 / (base ** (torch.arange(0, dim, 2).float() / dim))
        t = torch.arange(max_seq_len, dtype=torch.float32)
        freqs = torch.outer(t, inv_freq)
        emb = torch.cat((freqs, freqs), dim=-1)
        self.register_buffer('cos_cached', emb.cos(), persistent=False)
        self.register_buffer('sin_cached', emb.sin(), persistent=False)
    def forward(self, q, k, seq_len):
        cos = self.cos_cached[:seq_len].unsqueeze(0).unsqueeze(1).to(q.dtype).to(q.device)
        sin = self.sin_cached[:seq_len].unsqueeze(0).unsqueeze(1).to(k.dtype).to(k.device)
        q1, q2 = q[..., :self.dim//2], q[..., self.dim//2:]
        k1, k2 = k[..., :self.dim//2], k[..., self.dim//2:]
        return (q * cos) + (torch.cat((-q2, q1), dim=-1) * sin), (k * cos) + (torch.cat((-k2, k1), dim=-1) * sin)

class DraftTransformer(nn.Module):
    def __init__(self, vocab_size=128256, hidden_dim=512, num_layers=5, num_heads=8, head_dim=64, intermediate_dim=1536):
        super().__init__()
        self.config = {'vocab_size': vocab_size, 'hidden_dim': hidden_dim, 'num_layers': num_layers, 'num_heads': num_heads, 'head_dim': head_dim, 'intermediate_dim': intermediate_dim, 'max_seq_len': 2048, 'rope_base': 500000.0, 'norm_eps': 1e-5, 'tie_word_embeddings': True}
        self.embed_tokens = nn.Embedding(vocab_size, hidden_dim)
        self.rope = RotaryEmbedding(head_dim)
        self.layers = nn.ModuleList([
            nn.ModuleDict({
                'norm1': RMSNorm(hidden_dim),
                'q': nn.Linear(hidden_dim, num_heads*head_dim, bias=False),
                'k': nn.Linear(hidden_dim, num_heads*head_dim, bias=False),
                'v': nn.Linear(hidden_dim, num_heads*head_dim, bias=False),
                'o': nn.Linear(num_heads*head_dim, hidden_dim, bias=False),
                'norm2': RMSNorm(hidden_dim),
                'gate': nn.Linear(hidden_dim, intermediate_dim, bias=False),
                'up': nn.Linear(hidden_dim, intermediate_dim, bias=False),
                'down': nn.Linear(intermediate_dim, hidden_dim, bias=False)
            }) for _ in range(num_layers)
        ])
        self.norm = RMSNorm(hidden_dim)
        self.lm_head = nn.Linear(hidden_dim, vocab_size, bias=False)
        self.lm_head.weight = self.embed_tokens.weight

    def forward(self, x):
        b, s = x.shape
        h = self.embed_tokens(x)
        for l in self.layers:
            n1 = l['norm1'](h)
            q, k, v = l['q'](n1).view(b,s,8,64).transpose(1,2), l['k'](n1).view(b,s,8,64).transpose(1,2), l['v'](n1).view(b,s,8,64).transpose(1,2)
            q, k = self.rope(q, k, s)
            attn = F.scaled_dot_product_attention(q, k, v, is_causal=True).transpose(1, 2).reshape(b, s, 512)
            h = h + l['o'](attn)
            n2 = l['norm2'](h)
            h = h + l['down'](F.silu(l['gate'](n2)) * l['up'](n2))
        return self.lm_head(self.norm(h))

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Training on: {device} ({torch.cuda.get_device_name(0) if torch.cuda.is_available() else "CPU"})')

print('Loading Meta Llama 3.1 Tokenizer...')
tokenizer = AutoTokenizer.from_pretrained('NousResearch/Meta-Llama-3.1-8B-Instruct')
if tokenizer.pad_token_id is None: tokenizer.pad_token_id = 128001

samples = []
print('Loading 70,000 Code & Instruction Programs...')

# 1. Magicoder (40,000 Python programs)
try:
    print('  [1/3] Loading Magicoder OSS Instruct (40,000 samples)...')
    m_data = load_dataset('ise-uiuc/Magicoder-OSS-Instruct-75K', split='train')
    for r in m_data:
        p = r.get('problem', r.get('instruction', ''))
        s = r.get('solution', r.get('response', ''))
        if p and s:
            toks = tokenizer.encode(f'<|start_header_id|>user<|end_header_id|>\n\n{p}<|eot_id|><|start_header_id|>assistant<|end_header_id|>\n\n{s}<|eot_id|>')
            if len(toks) > 20: samples.append(toks[:256])
        if len(samples) >= 40000: break
except Exception as e: print(f'Magicoder note: {e}')

# 2. Python Alpaca (18,000 Python programs)
try:
    print('  [2/3] Loading Python Code Instructions (18,000 samples)...')
    a_data = load_dataset('iamtarun/python_code_instructions_18k_alpaca', split='train')
    for r in a_data:
        p = r.get('instruction', '') + ('\n' + r.get('input', '') if r.get('input') else '')
        s = r.get('output', '')
        if p and s:
            toks = tokenizer.encode(f'<|start_header_id|>user<|end_header_id|>\n\n{p}<|eot_id|><|start_header_id|>assistant<|end_header_id|>\n\n{s}<|eot_id|>')
            if len(toks) > 20: samples.append(toks[:256])
        if len(samples) >= 58000: break
except Exception as e: print(f'Alpaca note: {e}')

# 3. Dolly (12,000 General QA & Reasoning)
try:
    print('  [3/3] Loading Dolly 15k (12,000 samples)...')
    d_data = load_dataset('databricks/databricks-dolly-15k', split='train')
    for r in d_data:
        toks = tokenizer.encode(f"<|start_header_id|>user<|end_header_id|>\n\n{r['instruction']}<|eot_id|><|start_header_id|>assistant<|end_header_id|>\n\n{r['response']}<|eot_id|>")
        if len(toks) > 16: samples.append(toks[:256])
        if len(samples) >= 70000: break
except Exception as e: print(f'Dolly note: {e}')

print(f'[OK] Collected {len(samples):,} tokenized programs (~{len(samples)*256/1e6:.1f} Million tokens)!')

def collate(b):
    ml = max(len(s) for s in b)
    inp, lbl = [], []
    for s in b:
        p = s + [128001]*(ml - len(s))
        inp.append(p[:-1])
        lbl.append([t if t!=128001 else -100 for t in p[1:]])
    return torch.tensor(inp, dtype=torch.long), torch.tensor(lbl, dtype=torch.long)

# Optimal Max-Speed Settings for Tesla T4 GPU
BATCH_SIZE = 16
GRAD_ACCUM_STEPS = 2

loader = DataLoader(samples, batch_size=BATCH_SIZE, shuffle=True, collate_fn=collate, pin_memory=True, num_workers=2)
model = DraftTransformer(vocab_size=len(tokenizer)).to(device)

if os.path.exists('draft_model.pt'):
    try:
        prev = torch.load('draft_model.pt', map_location=device)
        model.load_state_dict(prev['state_dict'], strict=False)
        print('[OK] Resumed from existing draft_model.pt checkpoint!')
    except Exception: pass

opt = torch.optim.AdamW(model.parameters(), lr=5e-4, weight_decay=0.01)
loss_fn = nn.CrossEntropyLoss(ignore_index=-100)
scaler = torch.amp.GradScaler('cuda')
epochs = 5
total_update_steps = (len(loader) // GRAD_ACCUM_STEPS) * epochs
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=total_update_steps, eta_min=1e-5)

print(f'Starting 70,000-Program GPU training for {epochs} epochs (Batch size 16, Grad Accum 2)...')
t0 = time.time()

for ep in range(1, epochs + 1):
    ep_loss = 0
    opt.zero_grad()
    for st, (x, y) in enumerate(loader):
        x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
        
        with torch.amp.autocast('cuda'):
            logits = model(x)
            loss = loss_fn(logits.view(-1, len(tokenizer)), y.view(-1))
            loss = loss / GRAD_ACCUM_STEPS

        scaler.scale(loss).backward()
        ep_loss += loss.item() * GRAD_ACCUM_STEPS

        if (st + 1) % GRAD_ACCUM_STEPS == 0 or (st + 1) == len(loader):
            scaler.unscale_(opt)
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scaler.step(opt)
            scaler.update()
            opt.zero_grad()
            scheduler.step()

        if (st + 1) % 500 == 0:
            print(f'Epoch [{ep}/{epochs}] Step [{st+1}/{len(loader)}] - Loss: {loss.item() * GRAD_ACCUM_STEPS:.4f}')

    avg = ep_loss / len(loader)
    print(f'==> [Epoch {ep}/{epochs}] Average Loss: {avg:.4f} | Perplexity: {math.exp(min(avg, 20)):.2f}')
    
    payload = {'config': model.config, 'state_dict': model.state_dict(), 'tokenizer_name': 'NousResearch/Meta-Llama-3.1-8B-Instruct', 'vocab_size': len(tokenizer), 'params_count': sum(p.numel() for p in model.parameters())}
    torch.save(payload, 'draft_model.pt')
    print(f'[Saved] Epoch {ep} checkpoint secured as draft_model.pt')

elapsed_min = (time.time() - t0) / 60
print(f'[OK] 70,000-Program Training Complete in {elapsed_min:.1f} minutes!')

# Auto-download final model
from google.colab import files
files.download('draft_model.pt')
```
